```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_openml
import pandas as pd
```

```python
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X, y = mnist.data, mnist.target.astype(int)
print("X shape:", X.shape)
print("y shape:", y.shape)
print("Pixel value range:", X.min(), "to", X.max())
```

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for digit in range(10):
    idx = np.where(y == digit)[0][0]
    ax = axes[digit // 5][digit % 5]
    ax.imshow(X[idx].reshape(28, 28), cmap='gray')
    ax.set_title(f"Label: {digit}")
    ax.axis('off')
plt.suptitle("Sample Digits from MNIST")
plt.tight_layout()
plt.savefig("../data/sample_digits.png")
plt.show()
```

```python
unique, counts = np.unique(y, return_counts=True)
plt.figure(figsize=(10, 4))
sns.barplot(x=unique, y=counts, palette="viridis")
plt.title("Class Distribution in MNIST")
plt.xlabel("Digit")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig("../data/class_distribution.png")
plt.show()
```

```python
print(f"Total samples: {X.shape[0]}")
print(f"Features per sample: {X.shape[1]}")
print(f"Classes: {np.unique(y)}")
print(f"Samples per class (min/max): {counts.min()} / {counts.max()}")
```
